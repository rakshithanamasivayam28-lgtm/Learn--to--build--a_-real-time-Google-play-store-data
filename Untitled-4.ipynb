```python
import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from datetime import datetime
import pytz
import re

# ---------------- Load Data ----------------
df = pd.read_csv("Play Store Data.csv")
df.columns = df.columns.str.strip()

# ---------------- Create Missing Columns ----------------
if 'Installs' not in df.columns:
    df['Installs'] = df.get('Total Installs', 0)

if 'Category' not in df.columns:
    df['Category'] = df.get('Genres', 'Other')

# ---------------- Cleaning ----------------
df['Installs'] = df['Installs'].astype(str)
df['Installs'] = df['Installs'].str.replace('+','', regex=False)
df['Installs'] = df['Installs'].str.replace(',','', regex=False)
df['Installs'] = pd.to_numeric(df['Installs'], errors='coerce')

df['Reviews'] = pd.to_numeric(df['Reviews'], errors='coerce')

df['Size'] = df['Size'].astype(str)
df['Size'] = df['Size'].str.replace('M','', regex=False)
df['Size'] = pd.to_numeric(df['Size'], errors='coerce')

df['Last Updated'] = pd.to_datetime(df['Last Updated'], errors='coerce')

# ---------------- Filters ----------------
df = df[
    (df['Rating'] >= 4.2) &
    (df['Reviews'] > 1000) &
    (df['Size'] >= 20) &
    (df['Size'] <= 80)
]

# App name must not contain numbers
df = df[df['App'].apply(lambda x: not bool(re.search(r'\d', str(x))))]

# Category must start with T or P
df = df[df['Category'].str.startswith(('T','P'))]

# ---------------- Group by Month ----------------
df['Month'] = df['Last Updated'].dt.to_period('M').astype(str)

monthly = df.groupby(['Month','Category'])['Installs'].sum().reset_index()
pivot = monthly.pivot(index='Month', columns='Category', values='Installs').fillna(0)

# ---------------- Cumulative ----------------
cumulative = pivot.cumsum()

# ---------------- Growth Highlight ----------------
growth = cumulative.pct_change() * 100
highlight_months = growth.max(axis=1) > 25

# ---------------- Legend Translation ----------------
legend_map = {
    "Travel & Local": "Voyage et Local",
    "Productivity": "Productividad",
    "Photography": "写真"
}

cumulative.rename(columns=legend_map, inplace=True)

# ---------------- Time Gate ----------------
ist = pytz.timezone("Asia/Kolkata")
current_time = datetime.now(ist).time()
start = datetime.strptime("16:00", "%H:%M").time()
end   = datetime.strptime("18:00", "%H:%M").time()

st.title("Cumulative Installs by Category")

# ---------------- Display ----------------
if start <= current_time <= end:

    fig, ax = plt.subplots(figsize=(10,6))

    cumulative.plot.area(ax=ax, alpha=0.7)

    for i, month in enumerate(cumulative.index):
        if highlight_months.loc[month]:
            ax.axvspan(i-0.5, i+0.5, color='red', alpha=0.15)

    ax.set_title("Stacked Area: Cumulative Installs Over Time")
    ax.set_xlabel("Month")
    ax.set_ylabel("Cumulative Installs")
    ax.legend(title="Category")

    st.pyplot(fig)

else:
    st.warning("This visualization is available only between 4 PM and 6 PM IST.")
```
