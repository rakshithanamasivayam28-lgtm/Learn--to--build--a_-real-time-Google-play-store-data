```python
import streamlit as st
import pandas as pd
import plotly.express as px
from datetime import datetime
import pytz
from textblob import TextBlob
import re

# ---------------- Load Data ----------------
df = pd.read_csv("Play Store Data.csv")
df.columns = df.columns.str.strip()

# ---------------- Create Missing Columns ----------------
if 'Installs' not in df.columns:
    df['Installs'] = df.get('Total Installs', 0)

if 'Category' not in df.columns:
    df['Category'] = df.get('Genres', 'Other')

# ---------------- Cleaning ----------------
df['Installs'] = df['Installs'].astype(str)
df['Installs'] = df['Installs'].str.replace('+','', regex=False)
df['Installs'] = df['Installs'].str.replace(',','', regex=False)
df['Installs'] = pd.to_numeric(df['Installs'], errors='coerce')

df['Reviews'] = pd.to_numeric(df['Reviews'], errors='coerce')

df['Size'] = df['Size'].astype(str)
df['Size'] = df['Size'].str.replace('M','', regex=False)
df['Size'] = pd.to_numeric(df['Size'], errors='coerce')

# ---------------- Sentiment Subjectivity ----------------
df['Subjectivity'] = df['App'].apply(lambda x: TextBlob(str(x)).sentiment.subjectivity)

# ---------------- Filters ----------------
valid_categories = [
    'Game','Beauty','Business','Comics',
    'Communication','Dating',
    'Entertainment','Social','Events'
]

df = df[
    (df['Rating'] > 3.5) &
    (df['Reviews'] > 500) &
    (df['Installs'] > 50000) &
    (df['Category'].isin(valid_categories)) &
    (df['Subjectivity'] > 0.5)
]

# App name must NOT contain "S"
df = df[~df['App'].str.contains('S', case=False, na=False)]

# ---------------- Translate Categories ----------------
translation_map = {
    "Beauty": "सुंदरता",
    "Business": "வணிகம்",
    "Dating": "Verabredung"
}

df['Category_Display'] = df['Category'].replace(translation_map)

# ---------------- Color Highlight ----------------
df['Color'] = df['Category'].apply(
    lambda x: 'pink' if x == 'Game' else 'blue'
)

# ---------------- IST Time Gate ----------------
ist = pytz.timezone("Asia/Kolkata")
current_time = datetime.now(ist).time()

start = datetime.strptime("17:00", "%H:%M").time()
end   = datetime.strptime("19:00", "%H:%M").time()

st.title("Bubble Chart: App Size vs Rating")

# ---------------- Display ----------------
if start <= current_time <= end:

    st.success("Chart is live (5 PM – 7 PM IST)")

    fig = px.scatter(
        df,
        x="Size",
        y="Rating",
        size="Installs",
        color="Category_Display",
        hover_name="App",
        title="Bubble Chart: App Size vs Rating",
        labels={
            "Size":"App Size (MB)",
            "Rating":"Average Rating",
            "Category_Display":"Category"
        }
    )

    # Force Game to pink
    for trace in fig.data:
        if "Game" in trace.name:
            trace.marker.color = "pink"

    st.plotly_chart(fig, use_container_width=True)

else:
    st.warning("This chart is available only between 5 PM and 7 PM IST.")
```
