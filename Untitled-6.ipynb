```python
import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime
import pytz
import re

# ---------------- Load Data ----------------
df = pd.read_csv("Play Store Data.csv")
df.columns = df.columns.str.strip()

# ---------------- Create Missing Columns ----------------
if 'Installs' not in df.columns:
    df['Installs'] = df.get('Total Installs', 0)

if 'Category' not in df.columns:
    df['Category'] = df.get('Genres', 'Other')

# ---------------- Cleaning ----------------
df['Installs'] = df['Installs'].astype(str)
df['Installs'] = df['Installs'].str.replace('+','', regex=False)
df['Installs'] = df['Installs'].str.replace(',','', regex=False)
df['Installs'] = pd.to_numeric(df['Installs'], errors='coerce')

df['Reviews'] = pd.to_numeric(df['Reviews'], errors='coerce')
df['Last Updated'] = pd.to_datetime(df['Last Updated'], errors='coerce')

# ---------------- Filters ----------------

# Reviews > 500
df = df[df['Reviews'] > 500]

# App name must not start with X,Y,Z
df = df[~df['App'].str.match('^[XYZxyz]', na=False)]

# App name must not contain S
df = df[~df['App'].str.contains('S', case=False, na=False)]

# Category must start with E, C, B
df = df[df['Category'].str.startswith(('E','C','B'))]

# ---------------- Group by Month ----------------
df['Month'] = df['Last Updated'].dt.to_period('M').astype(str)

monthly = df.groupby(['Month','Category'])['Installs'].sum().reset_index()
pivot = monthly.pivot(index='Month', columns='Category', values='Installs').fillna(0)

# ---------------- Growth Detection ----------------
growth = pivot.pct_change() * 100
highlight_months = growth.max(axis=1) > 20

# ---------------- Translate Legend ----------------
translation_map = {
    "Beauty": "सुंदरता",
    "Business": "வணிகம்",
    "Dating": "Verabredung"
}
pivot.rename(columns=translation_map, inplace=True)

# ---------------- Time Gate ----------------
ist = pytz.timezone("Asia/Kolkata")
current_time = datetime.now(ist).time()

start = datetime.strptime("18:00", "%H:%M").time()
end   = datetime.strptime("21:00", "%H:%M").time()

st.title("Time Series: Total Installs by Category")

# ---------------- Display ----------------
if start <= current_time <= end:

    fig, ax = plt.subplots(figsize=(10,6))

    pivot.plot(ax=ax, marker='o')

    for i, month in enumerate(pivot.index):
        if highlight_months.loc[month]:
            ax.axvspan(i-0.5, i+0.5, color='orange', alpha=0.2)

    ax.set_title("Trend of Total Installs Over Time")
    ax.set_xlabel("Month")
    ax.set_ylabel("Total Installs")
    ax.legend(title="Category")

    st.pyplot(fig)

else:
    st.warning("This visualization is available only between 6 PM and 9 PM IST.")
```
