```python
import streamlit as st
import pandas as pd
import plotly.express as px
from datetime import datetime
import pytz

# ------------------ Load Data ------------------
df = pd.read_csv("User Reviews.csv")

# Remove hidden spaces
df.columns = df.columns.str.strip()

# ------------------ CREATE CATEGORY ------------------
if 'Category' not in df.columns:
    if 'Genres' in df.columns:
        df['Category'] = df['Genres']
    elif 'Type' in df.columns:
        df['Category'] = df['Type']
    else:
        df['Category'] = 'Other'

# ------------------ CREATE INSTALLS ------------------
if 'Installs' not in df.columns:
    if 'Total Installs' in df.columns:
        df['Installs'] = df['Total Installs']
    elif 'Install' in df.columns:
        df['Installs'] = df['Install']
    elif 'Downloads' in df.columns:
        df['Installs'] = df['Downloads']
    else:
        df['Installs'] = 0

# ------------------ CLEAN INSTALLS ------------------
df['Installs'] = df['Installs'].astype(str)
df['Installs'] = df['Installs'].str.replace('+','', regex=False)
df['Installs'] = df['Installs'].str.replace(',','', regex=False)
df['Installs'] = pd.to_numeric(df['Installs'], errors='coerce')

# ------------------ CATEGORY FILTER ------------------
df = df[~df['Category'].astype(str).str.startswith(('A','C','G','S'))]

# ------------------ TOP 5 CATEGORIES ------------------
top_categories = (
    df.groupby('Category')['Installs']
    .sum()
    .sort_values(ascending=False)
    .head(5)
    .index
)

df = df[df['Category'].isin(top_categories)]

# ------------------ HIGHLIGHT ------------------
df['Highlight'] = df['Installs'].apply(
    lambda x: 'Above 1M Installs' if x > 1_000_000 else 'Below 1M Installs'
)

# ------------------ CREATE COUNTRY ------------------
if 'Country' not in df.columns:
    df['Country'] = 'India'

# ------------------ AGGREGATE ------------------
country_data = (
    df.groupby(['Country', 'Category', 'Highlight'])['Installs']
    .sum()
    .reset_index()
)

# ------------------ TIME GATE ------------------
ist = pytz.timezone("Asia/Kolkata")
current_time = datetime.now(ist).time()

start = datetime.strptime("18:00", "%H:%M").time()
end   = datetime.strptime("20:00", "%H:%M").time()

st.title("Global App Installs – Choropleth Map")

# ------------------ DISPLAY ------------------
if start <= current_time <= end:

    st.success("Map is live (6 PM – 8 PM IST)")

    fig = px.choropleth(
        country_data,
        locations="Country",
        locationmode="country names",
        color="Installs",
        hover_name="Category",
        hover_data=["Highlight"],
        animation_frame="Category",
        color_continuous_scale="Plasma",
        title="Global Installs by Top 5 App Categories"
    )

    st.plotly_chart(fig, use_container_width=True)

else:
    st.warning("This visualization is available only between 6 PM and 8 PM IST.")
```
